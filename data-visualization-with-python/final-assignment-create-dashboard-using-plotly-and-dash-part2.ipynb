## TASK 2.4: Creating Callbacks; Define the callback function to update the input container based on the selected statistics and the output container

### Update Input Container callback function
The purpose of this function is to enable or disable the year selection dropdown based on the user's choice of report type from another dropdown. Here we are having 2 parts:

**Callback definition**
The parameters of the Callback definition are as follows:
- Output: Specifies the output of the callback, which in this case is the disabled property of the dropdown with the ID select-year.
- Input: Specifies the input that triggers the callback, which is the value property of the dropdown with the ID dropdown-statistics.

**Callback Function**
The function checks the value of selected statistics. If the selected statistics is Yearly Statistics the function returns False. This means the disabled property of the select-year dropdown will be set to False, enabling the dropdown so the user can select a year.

If selected statistics is not Yearly Statistics (i.e., it is Recession Period Statistics), the function returns True. This means the disabled property of the select-year dropdown will be set to True, disabling the dropdown as selecting a year is not relevant in this context.

### Update Output Container callback function
Our layout has 1 output container, and we will be required to return the plots developed dcc.Graph() into this container as divisions.
For each report we need to display four plots.

The update_output_container function in this Dash application dynamically generates and displays graphs based on the user's selection from the dropdown menus. Specifically, it creates different sets of graphs for Recession Period Statistics and Yearly Statistics. We will be creating 4 plots for each report.

Here we have 2 parts:

**Callback Definition**
Parameters:
- Output: This specifies that the children property of the component with the ID output-container will be updated.In Dash, children typically refers to the content within a container, such as graphs, text, or other HTML elements.
- Input: Input Property for Report type:This specifies that the value property of the dropdown menu with the ID dropdown-statistics (which allows the user to select between Yearly Statistics and Recession Period Statistics) will trigger the callback when it changes. Input property for year selection: This specifies that the value property of the dropdown menu with the ID select-year (which allows the user to select a specific year) will trigger the callback when it changes.

**Callback Function**
Here you will filter the data based on the report type selected.

When Recession Period Statistics is selected, the data is filtered to include only recession periods where Recession equals 1. Conversely, when Yearly Statistics is chosen, the data is filtered based on the selected year.

## Complete code

```python
import requests
import io
from dash import Dash, html, dcc, Input, Output
import plotly.express as px
import pandas as pd
from dash.dependencies import Input, Output

URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"

response = requests.get(URL)
response.raise_for_status()
csv_content = io.StringIO(response.text)
df = pd.read_csv(csv_content)

app = Dash(__name__)
# List of years 
year_list = [i for i in range(1980, 2024)]

app.layout = html.Div([
    html.H1("Automobile Sales Statistics Dashboard", style={'textAlign':'center', 'color':'#503D36', 'font-size':'24px', 'font-weight': 'bold'}),
    
    # Create a dropdown menu and add two Report options to it.
    html.Div([
        dcc.Dropdown(
            id='dropdown-statistics', # Set id to dropdown-statistics
            options=[ # Set options to a list containing dictionaries with label and user-provided values for value.
                    {'label': 'Yearly Statistics', 'value': 'Yearly Statistics'},
                    {'label': 'Recession Period Statistics', 'value': 'Recession Period Statistics'}
                ],
            value='Select Statistics', # Set value to Select Statistics.
            placeholder='Select a report type', # Set placeholder to Select a report type.
            style={'width': '60%', 'padding': '3px','font-size': '20px', 'text-align-last':'center', 'margin-left':'20%'}
        )
    ]),
    # Create a dropdown menu for selecting the YEAR options to it.
    html.Div([
        dcc.Dropdown(
            id='dropdown-year', # Set id to select-year
            options=[{'label': i, 'value': i} for i in year_list], # list of years as options in this dropdowwn
            placeholder='Select a year', # Set placeholder to Select-year
            style={'width': '60%', 'padding': '3px','font-size': '20px', 'text-align-last':'center', 'margin-left':'20%', 'margin-bottom':'20px'}
        )
    ]),
    # Add a division for output display with appropriate id and classname property
    html.Div([
        html.Div(
            id='output-container', 
            className='chart-dashboard', 
            style={'width': '100%','font-size': '20px', 'text-align-last':'center'}
        )
    ])
])

# Callback definition
# - Output: Specifies the output of the callback, which in this case is the disabled property of the dropdown with the ID select-year.
# - Input: Specifies the input that triggers the callback, which is the value property of the dropdown with the ID dropdown-statistics.
@app.callback(
    Output(component_id='dropdown-year', component_property='disabled'),
    Input(component_id='dropdown-statistics', component_property='value')
)

def update_dropdown_year_status(selected_statistics):
    '''
    The function checks the value of selected statistics. 
    If the selected statistics is Yearly Statistics the function returns False. 
    If selected statistics is not Yearly Statistics, 
    the function returns True. 
    '''
    if selected_statistics == 'Yearly Statistics': 
        return False
    else: 
        return True

# Update the dropdown-year's value property 
@app.callback(
    Output(component_id='dropdown-year', component_property='value'),
    Input(component_id='dropdown-statistics', component_property='value')
)
def update_dropdown_year_value(selected_statistics):
    if selected_statistics == 'Recession Period Statistics': 
        return ""

# Callback Definition
# - Output: This specifies that the children property of the component with the ID output-container will be updated.In Dash, children typically refers to the content within a container, such as graphs, text, or other HTML elements.
# - Input: Input Property for Report type:This specifies that the value property of the dropdown menu with the ID dropdown-statistics (which allows the user to select between Yearly Statistics and Recession Period Statistics) will trigger the callback when it changes. 
# - Input property for year selection: This specifies that the value property of the dropdown menu with the ID select-year (which allows the user to select a specific year) will trigger the callback when it changes.
@app.callback(
    Output(component_id='output-container', component_property='children'),
    [Input(component_id='dropdown-statistics', component_property='value'), Input(component_id='dropdown-year', component_property='value')]
)

# Update Output Container callback function
def update_output_container(selected_statistics, input_year):
    '''
    Filter the data based on the report type selected.
    '''
    if selected_statistics == 'Recession Period Statistics':

        # When Recession Period Statistics is selected, the data is filtered to include only recession periods where Recession equals 1. 
        recession_data = df[df['Recession'] == 1]
        
        # Plot 1. Automobile sales fluctuate over Recession Period 
        yearly_recession = recession_data.groupby('Year')['Automobile_Sales'].mean().reset_index()
        chart1 = dcc.Graph(
            figure=px.line(yearly_recession,
            x='Year',
            y='Automobile_Sales',
            title="Average Automobile Sales fluctuation")
        )

        # Plot 2. Calculate the average number of vehicles sold by vehicle type       
        average_sales = recession_data.groupby('Vehicle_Type')['Automobile_Sales'].mean().reset_index()                           
        chart2 = dcc.Graph(
            figure=px.bar(average_sales,
            x='Vehicle_Type',
            y='Automobile_Sales',
            title="Average Automobile Sales by Vehicle Type")
        )
    
        # Plot 3. Expenditure share by vehicle type during recessions
        advert_expenditure = recession_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
        chart3 = dcc.Graph(
            figure=px.pie(advert_expenditure,
            values='Advertising_Expenditure',
            names='Vehicle_Type',
            title="Expenditures by Vehicle Type")
        )

        # Plot 4. Effect of unemployment rate on vehicle type and sales
        unemp_rate = recession_data.groupby('Vehicle_Type')['unemployment_rate'].mean().reset_index()
        chart4 = dcc.Graph(
            figure=px.bar(unemp_rate,
            x='Vehicle_Type',
            y='unemployment_rate',
            title="Unemployment Rate on Automobile Type and Sales")
        )      

        # Returning the graphs for displaying recession periods data
        return [
            html.Div(className='chart-item', children=[html.Div(children=chart1),html.Div(children=chart2)],style={'display': 'flex', 'justify-content': 'center'}),
            html.Div(className='chart-item', children=[html.Div(children=chart3),html.Div(children=chart4)],style={'display': 'flex', 'justify-content': 'center'})
        ]                          
                
    # When Yearly Statistics is chosen, the data is filtered based on the selected year.
    elif (input_year and selected_statistics == 'Yearly Statistics'):
        
        # Filter the data for selected year
        yearly_data = df[df['Year'] == input_year]
                
        # Plot 1. Yearly Automobile sales.
        automobile_sales = df.groupby('Year')['Automobile_Sales'].sum().reset_index()
        chart1 = dcc.Graph(
            figure=px.line(automobile_sales,
            x='Year',
            y='Automobile_Sales',
            title=f"Yearly Automobile sales - {input_year}")
        )
            
        # Plot 2. Monthly Automobile sales.
        monthly_sales = yearly_data.groupby('Month')['Automobile_Sales'].sum().reset_index()
        chart2 = dcc.Graph(
            figure=px.line(monthly_sales,
            x='Month',
            y='Automobile_Sales',
            title=f"Monthly Automobile sales - {input_year}")
        )

        # Plot 3. Average number of vehicles sold for selected year
        average_aut_sales = yearly_data.groupby('Year')['Automobile_Sales'].mean().reset_index()
        chart3 = dcc.Graph(
            figure=px.bar(average_aut_sales,
            x='Year',
            y='Automobile_Sales',
            range_x=[input_year, input_year],
            title=f"Average Vehicles Sold by Vehicle Type - {input_year}")
        )

        # Plot 3. Advertisement Expenditure for each vehicle
        advert_expenditure = yearly_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum().reset_index()
        chart4 = dcc.Graph(
            figure=px.pie(advert_expenditure,
            values='Advertising_Expenditure',
            names='Vehicle_Type',
            title=f"Total Expenditures by Vehicle Type - {input_year}")
        )
            
        # Returning the graphs for displaying Yearly data
        return [
            html.Div(className='chart-item', children=[html.Div(children=chart1),html.Div(children=chart2)],style={'display': 'flex', 'justify-content': 'center'}),
            html.Div(className='chart-item', children=[html.Div(children=chart3),html.Div(children=chart4)],style={'display': 'flex', 'justify-content': 'center'})
        ]
    else:
        return None
    
if __name__ == '__main__':
    app.run_server(debug=False)
```
